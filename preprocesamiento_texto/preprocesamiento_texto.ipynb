# Preprocesamiento de texto en español

Actividad basada en los códigos fuente de los slides de la semana 7: limpieza, tokenización, stemming, lematización, stopwords y vectorización TF-IDF.

```python
!pip -q install nltk spacy scikit-learn
!python -m spacy download es_core_news_sm

import re
import nltk
import spacy
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
from nltk.tokenize import sent_tokenize, word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nlp = spacy.load("es_core_news_sm")
```

## Texto de trabajo

```python
texto = (
    "Me encanta aprender procesamiento de lenguaje natural. "
    "El procesamiento de textos ayuda a que los modelos de inteligencia artificial "
    "comprendan mejor la información."
)
print(texto)
```

## 1. Limpieza y tokenización

Se convierten las palabras a minúsculas y se eliminan URLs, números y caracteres especiales.

```python
texto_limpio = texto.lower()
texto_limpio = re.sub(r"https?://\S+|www\.\S+", "", texto_limpio)
texto_limpio = re.sub(r"[^a-záéíóúüñ\s]", " ", texto_limpio)
texto_limpio = re.sub(r"\s+", " ", texto_limpio).strip()

tokens_palabra = word_tokenize(texto_limpio, language="spanish")
tokens_oracion = sent_tokenize(texto, language="spanish")
print("Texto limpio:", texto_limpio)
print("Tokens de palabras:", tokens_palabra)
print("Tokens de oraciones:", tokens_oracion)
```

## 2. Eliminación de stopwords

```python
stop_words = set(stopwords.words("spanish"))
tokens_sin_stopwords = [
    token for token in tokens_palabra if token not in stop_words
]
print("Tokens sin stopwords:", tokens_sin_stopwords)
```

## 3. Stemming

El stemming reduce las palabras a una raíz aproximada. Puede producir raíces que no son palabras completas.

```python
stemmer = SnowballStemmer("spanish")
stems = [stemmer.stem(token) for token in tokens_sin_stopwords]
print("Stemming:", stems)
```

## 4. Lematización

La lematización obtiene la forma base real de una palabra considerando el idioma.

```python
doc = nlp(texto_limpio)
lemas = [
    token.lemma_
    for token in doc
    if token.is_alpha and not token.is_stop
]
print("Lemas:", lemas)
```

## 5. Vectorización con TF-IDF

TF-IDF transforma el texto en valores numéricos que pueden utilizarse como entrada de un modelo.

```python
vectorizador = TfidfVectorizer(
    lowercase=True,
    stop_words=list(stop_words),
)
matriz_tfidf = vectorizador.fit_transform([texto_limpio])
print("Vocabulario:", vectorizador.get_feature_names_out())
print("Matriz TF-IDF:")
print(matriz_tfidf.toarray())
```
