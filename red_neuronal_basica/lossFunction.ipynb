```python
import random

entradas = [-1,-2,-3, 1, 2, 3]
salida = [0, 0, 0, 1, 1, 1]


pesoActual = 0
pesoNuevo = 0
sesgo = 0
tazaDeAprendizaje = 0.1


def SumaPonderada(entradas, peso, sesgo):
      return (entradas * peso) + sesgo


def FuncionActivacion(z):
    if z >= 0:
        return 1
    else:
        return 0

def funcionPerdida(salida_esperada, salida_obtenida):
    return (salida_esperada - salida_obtenida) ** 2


def pesoNuevo(pesoActual, tazaDeAprendizaje, salida, entrada, prediccion):

    return pesoActual + (tazaDeAprendizaje * (salida - prediccion) * entrada)

def sesgoNuevo(sesgoActual, tazaDeAprendizaje, salida, prediccion):
    return sesgoActual + (tazaDeAprendizaje * (salida - prediccion))



for x in entradas:
    z = SumaPonderada(x, peso, sesgo)
    salida_obtenida = FuncionActivacion(z)
    perdida = funcionPerdida(salida[entradas.index(x)], salida_obtenida)
    print(f"Entrada: {x}, Salida obtenida: {salida_obtenida}, Salida esperada: {salida[entradas.index(x)]}, Pérdida: {perdida}")
    peso = pesoNuevo(peso, tazaDeAprendizaje, salida[entradas.index(x)], x, salida_obtenida)
    sesgo = sesgoNuevo(sesgo, tazaDeAprendizaje, salida[entradas.index(x)], salida_obtenida)
```
