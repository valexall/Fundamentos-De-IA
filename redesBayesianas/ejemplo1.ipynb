# Ejemplo de red bayesiana

Simulación para determinar si o no llueve y cual es el pronostico de lluvia. y la probalidad de llevar accesorio de lluvia

      Lluvia
/               \
pronostico    accesoria de lluvia

```python
#!pip install pgmpy
```

```python
# libreria de redes bayesianas
# crear la red bayesiana
from pgmpy.models import DiscreteBayesianNetwork
# crear una tabla de probabilidades condicionales
from pgmpy.factors.discrete import TabularCPD
# realizar inferencia en la red bayesiana
from pgmpy.inference import VariableElimination



```

```python
# crear la red bayesiana
model = DiscreteBayesianNetwork([
    ("Lluvia", "Pronostico"),
    ("Lluvia", "AccesorioLluvia") # accesorio de lluvia puede ser paraguas, impermeable, plastico, etc.
])
```

```python
# crear la tabla de probabilidades condicionales para la variable Lluvia

# que puede 30% de probabilidad de que llueva y 70% de probabilidad de que no llueva

cpd_lluvia = TabularCPD(
    variable="Lluvia",
    variable_card=2,
    values=[[0.3], [0.7]],
    state_names={"Lluvia": ["Sí", "No"]}
)
```

```python
# crear la tabla de probabilidades para la variable Pronostico
# es un hipotesis de incertidumbre, 
# que puede 80% de probabilidad de que el pronóstico sea "Sí" si llueve 
# y 20% de probabilidad de que el pronóstico sea "No" si llueve
cpd_pronostico = TabularCPD(
    variable="Pronostico",
    variable_card=2,
    values=[[0.8, 0.2], [0.2, 0.8]],
    evidence=["Lluvia"],
    evidence_card=[2],
    state_names={"Pronostico": ["Sí", "No"], "Lluvia": ["Sí", "No"]}
)
```

```python
# crear la tabla de probabilidades condicionales para accesorio de lluvia
# que puede 90% de probabilidad de que se lleve un accesorio de lluvia
# si llueve y 10% de probabilidad de que se lleve un accesorio de lluvia si no llueve
cpd_accesorio_lluvia = TabularCPD(
    variable="AccesorioLluvia",
    variable_card=2,
    values=[[0.9, 0.1], [0.1, 0.9]],
    evidence=["Lluvia"],
    evidence_card=[2],
    state_names={"AccesorioLluvia": ["Sí", "No"], "Lluvia": ["Sí", "No"]}
)   
```

```python
# agregar al modelo las tablas de probabilidades condicionales
model.add_cpds(cpd_lluvia, cpd_pronostico, cpd_accesorio_lluvia)
model.check_model()
```

```python
print("Modelo de red bayesiana creado y verificado correctamente.")
print(model.edges())
```

```python
# Inferencia en la red bayesiana4
# motor o proceso de inferencia para al grafo
inferencia = VariableElimination(model)
resultado_lluvia = inferencia.query(
    variables=["Lluvia"],
    evidence={"Pronostico": "Sí", "AccesorioLluvia": "Sí"}
)
print("Probabilidad de que llueva dado que el pronóstico es 'Sí' y se lleva un accesorio de lluvia:")
print(resultado_lluvia)
```

```python
resultado_lluvia_2 = inferencia.query(
    variables=["Lluvia"])

print("Probabilidad de que llueva sin evidencia:")
print(resultado_lluvia_2)
```
